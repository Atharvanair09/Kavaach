# **Model Modeling**

### **Tokenizing English sentences**

```python
from transformers import DistilBertTokenizer

tokenizer = DistilBertTokenizer.from_pretrained("distilbert-base-uncased")
```

```python
def convert_tokenizer(example):
    return tokenizer(
        example["Text"],
        truncation = True,
        padding = "max_length",
        max_length = 128
    )
```

```python
# !pip install datasets
```

```python
from datasets import Dataset

train_dataset = Dataset.from_pandas(df_train)
test_dataset = Dataset.from_pandas(df_test)
val_dataset = Dataset.from_pandas(df_val)
```

```python
train_dataset = train_dataset.map(
    convert_tokenizer,
    batched = True,
)

test_dataset = test_dataset.map(
    convert_tokenizer,
    batched = True,
)

val_dataset = val_dataset.map(
    convert_tokenizer,
    batched = True,
)
```

```python
print(train_dataset[0])
```

### **Converting basic english labels into a group of 1 and 0 to get a array** 

```python
emotion_labels = [
    "Fear",
    "Anxiety",
    "Panic",
    "Distress",
    "Urgency",
    "Coercion",
    "Aggression",
    "Helplessness",
    "Calm",
]

def create_labels(examples):
    examples["labels"] = [
        1.0* examples[label]
        for label in emotion_labels
    ]
    return examples
```

```python
train_dataset = train_dataset.map(create_labels)
test_dataset = test_dataset.map(create_labels)
val_dataset = val_dataset.map(create_labels)
```

### **Implementing Binary Cross Entropy(BCE)**

```python
from numpy import dtype
import torch

pos_weight = torch.tensor(weights, dtype = torch.float32)
pos_weight
```

```python
torch.nn.BCEWithLogitsLoss(
    pos_weight = pos_weight
)
```

```python
from transformers import Trainer

class WeightedTrainer(Trainer):
    def compute_loss(
        self,
        model,
        inputs,
        return_outputs = False,
        **kwargs,
    ):
        labels = inputs.pop("labels")
        outputs = model(**inputs)

        loss_fct = torch.nn.BCEWithLogitsLoss(
            pos_weight = pos_weight.to(outputs.logits.device)
        )

        loss = loss_fct(
            outputs.logits,
            labels.float()
        )

        return ((loss, outputs) if return_outputs else loss)
```

### **Establishing the model**

```python
from transformers import DistilBertForSequenceClassification

model = DistilBertForSequenceClassification.from_pretrained(
    "distilbert-base-uncased",
    num_labels = 9,
    problem_type = "multi_label_classification"
)
```

### **Preparing the Evaluation for the model**

```python
from sklearn.metrics import (
    f1_score,
    precision_score,
    recall_score,
)

import numpy as np
import torch
```

```python
# !pip install transformers[torch]
```

```python
from transformers import TrainingArguments

training_arguments = TrainingArguments(
    output_dir = "C:/Users/Atharva Nair/kavaach/ml-model/model/kavaach",
    learning_rate = 2e-5,
    per_device_train_batch_size = 16,
    per_device_eval_batch_size = 16,
    num_train_epochs = 3,
)
```

```python
trainer = WeightedTrainer(
    model = model,
    args = training_arguments,
    train_dataset = train_dataset,
    eval_dataset= test_dataset
)
```

### **Starting the training**

```python
trainer.train()
```
