```python
from datasets import load_dataset
```

```python
emotions = load_dataset("go_emotions")

test = emotions["test"].to_pandas()
train = emotions["train"].to_pandas()
validation = emotions["validation"].to_pandas()
```

```python
import pandas as pd
combined_df = pd.concat([train, validation, test], ignore_index=True)
```

```python
combined_df.to_csv("dataset.csv", index=False)
```
